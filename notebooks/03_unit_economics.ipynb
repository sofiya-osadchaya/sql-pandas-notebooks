# Observed revenue and hypothetical CAC/LTV

демо-реконструкция подхода на синтетических данных

Use RETAIL_DATASET for the public UCI source; otherwise the run explicitly uses fictional data. See DATA_SOURCES.md.


```python
import os, sys
from pathlib import Path
root = Path.cwd() if (Path.cwd()/'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root))
from src.data import load_transactions
transactions, source_label = load_transactions(os.getenv('RETAIL_DATASET'))
print(source_label)
print('Rows:', len(transactions))
```

```python
observed=transactions.groupby('customer').revenue.sum()
print('Observed revenue per customer in the sample window:',round(observed.mean(),2))
# Explicit synthetic business assumptions; neither exists in the public source.
acquisition_spend=12000
new_customers=200
gross_margin=0.35
orders_per_year=3
assumed_lifetime_years=2
cac=acquisition_spend/new_customers
average_order=transactions.groupby('invoice').revenue.sum().mean()
ltv=average_order*orders_per_year*assumed_lifetime_years*gross_margin
print({'hypothetical_CAC':cac,'hypothetical_margin_LTV':round(ltv,2),'LTV_to_CAC':round(ltv/cac,2)})
assert cac>0 and ltv>0
```
